Import Library

```python
import pandas as pd
import warnings 
warnings.filterwarnings("ignore")
```

Load Data

```python
# Load data
df_detail_report = pd.read_csv("data/transactions.csv")
df_detail_report
```

```python
df_detail_report.info()
```

```python
df_detail_report['date'] = pd.to_datetime(df_detail_report['date'], format="%Y-%m-%d")
df_detail_report['date']
```

```python
# Find the last date in the DataFrame
last_date = df_detail_report['date'].max()

# Calculate three months before the last date
three_months_before = last_date - pd.DateOffset(months=3)

# Filter the DataFrame for dates three months before the last date
df_detail_report = df_detail_report[df_detail_report['date'] >= three_months_before]
df_detail_report
```

```python
# make dataframe to count quantity
df_quantity = df_detail_report[['menu', 'qty']]
df_quantity
```

```python
# count quantity
quantity = df_quantity.groupby(['menu']).sum().reset_index()
quantity
```

```python
df_menu = pd.read_csv("data/menu.csv")
df_menu
```

#### MINUMAN


```python
df_minuman = df_menu
df_minuman
```

```python
# merge df quantity and df_cocktails based on menu
df_minuman = df_minuman.merge(quantity, left_on='menu', right_on='menu', how='left')
df_minuman
```

```python
df_minuman = df_minuman[df_minuman['category'] == 'Beverage']
df_minuman
```

```python
df_minuman = df_minuman.drop(columns=['category', 'menu_sk', 'last_update', 'sub_category'])
df_minuman = df_minuman.rename(columns={'basic_cost':'Item Food Cost', 'price':'Item Sell Price', 'qty':'Number Sold'})
df_minuman
```

```python
item_count = df_minuman['menu'].value_counts().sum()
item_count
```

```python
df_minuman['Number Sold'] = df_minuman['Number Sold'].fillna(0).astype(int)
df_minuman
```

```python
# total number sold
sum_number_sold = df_minuman['Number Sold'].sum().astype(int)
sum_number_sold
```

```python
# kolom popularity in %
popularity = (df_minuman['Number Sold'] / sum_number_sold) * 100
popularity
```

```python
df_minuman['Popularity'] = popularity
df_minuman
```

```python
# item profit
df_minuman['Item Profit'] = df_minuman['Item Sell Price'] - df_minuman['Item Food Cost']
df_minuman
```

```python
# total cost
df_minuman['Total Cost'] = df_minuman['Number Sold'] * df_minuman['Item Food Cost']
df_minuman
```

```python
# Total revenue
df_minuman['Total Revenue'] = df_minuman['Number Sold'] * df_minuman['Item Sell Price']
df_minuman
```

```python
# total profit
df_minuman['Total Profit'] = df_minuman['Total Revenue'] - df_minuman['Total Cost']
df_minuman
```

```python
# rumus mencari I (sum total cost)
sum_total_cost = df_minuman['Total Cost'].sum()

# rumus mencari J (sum total revenue)
sum_total_revenue = df_minuman['Total Revenue'].sum()

# rumus mencari M (sum total profit)
sum_total_profit = df_minuman['Total Profit'].sum()

# rumus mencari O (average gross profit)
ave_gross_profit = sum_total_profit / sum_number_sold

# rumus mencari K (food cost %)
food_cost_k = (sum_total_cost / sum_total_revenue) * 100

# rumus mencari Q (menu popularity factor)
menu_popularity_factor = (1 / item_count) * 70/100

print(sum_total_cost)
print(sum_total_revenue)
print(sum_total_profit)
print(ave_gross_profit)
print(food_cost_k)
print(menu_popularity_factor)
```

```python
# Profit Category
df_minuman.loc[df_minuman['Item Profit'] < ave_gross_profit, 'Profit Category'] = 'Low'
df_minuman.loc[df_minuman['Item Profit'] > ave_gross_profit, 'Profit Category'] = 'High'
df_minuman
```

```python
# Popularity Category
df_minuman.loc[df_minuman['Popularity'] < menu_popularity_factor, 'Popularity Category'] = 'Low'
df_minuman.loc[df_minuman['Popularity'] > menu_popularity_factor, 'Popularity Category'] = 'High'
df_minuman
```

```python
# Popularity Category
df_minuman.loc[(df_minuman['Profit Category'] == 'High') & (df_minuman['Popularity Category'] == 'High'), 'Menu Item Class'] = 'STAR'
df_minuman.loc[(df_minuman['Profit Category'] == 'High') & (df_minuman['Popularity Category'] == 'Low'), 'Menu Item Class'] = 'PUZZLE'
df_minuman.loc[(df_minuman['Profit Category'] == 'Low') & (df_minuman['Popularity Category'] == 'High'), 'Menu Item Class'] = 'HORSE'
df_minuman.loc[(df_minuman['Profit Category'] == 'Low') & (df_minuman['Popularity Category'] == 'Low'), 'Menu Item Class'] = 'DOG'
df_minuman
```
